```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("train.csv")

df['SalePrice'] = pd.to_numeric(df['SalePrice'])
df['LotArea'] = pd.to_numeric(df['LotArea'])

df.dtypes
```

```python
df.isna().sum()
```

```python
df_missing = df.copy()
df_missing.loc[0:5, 'Amount'] = np.nan
df_missing.isna().sum()
```

```python
print("Original shape: ",df.shape)
print("After removing some values: ",df_missing.shape)
```

```python
df_missing.head(10)
```

```python
df_removed = df_missing.dropna()
df_removed.shape
```

```python
df_removed.isna().sum()
```

```python
df_missing.head(10)
```

```python
df_imputed_mean = df_missing.copy()
df_imputed_mean['Amount'].fillna(df_imputed_mean['Amount'].mean(), inplace=True)

df_imputed_mean.isna().sum()
```

```python
df_imputed_mean.head(10)
```

```python
df_imputed_median = df_missing.copy()
df_imputed_median['Amount'].fillna(df_imputed_median['Amount'].median(), inplace=True)

df_imputed_median.isna().sum()
```

```python
df_imputed_median.head(10)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['SalePrice'])
plt.title("Boxplot of House Sale Prices")
plt.xlabel("Sale Price")
plt.show()
```

```python
Q1 = df['SalePrice'].quantile(0.25)
Q3 = df['SalePrice'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['SalePrice'] < lower) | (df['SalePrice'] > upper)]
outliers.head(15)
```

```python
df_no_outliers = df[(df['SalePrice'] >= lower) & (df['SalePrice'] <= upper)]

print("Original shape: ", df.shape)
print("After removing outliers: ", df_no_outliers.shape)
```

```python
lower_cap = df['SalePrice'].quantile(0.05)
upper_cap = df['SalePrice'].quantile(0.95)

df_capped = df.copy()
df_capped['SalePrice'] = df_capped['SalePrice'].clip(lower_cap, upper_cap)
```

```python
df[['SalePrice', 'LotArea']].head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_scaled = df[['SalePrice', 'LotArea']].copy()

df_scaled[['SalePrice', 'LotArea']] = scaler.fit_transform(df_scaled)

df_scaled.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_standardized = df[['SalePrice', 'LotArea']].copy()

df_standardized[['SalePrice', 'LotArea']] = scaler.fit_transform(df_standardized)

df_standardized.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.heatmap(df_standardized[['SalePrice','LotArea']].corr(),
            annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap (Before PCA)")
plt.show()
```

```python
from sklearn.decomposition import PCA

X = df_standardized[['SalePrice', 'LotArea']]

pca = PCA(n_components=2)
principal_components = pca.fit_transform(X)

print("Explained Variance Ratio:", pca.explained_variance_ratio_)
```
