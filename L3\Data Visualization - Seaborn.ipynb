```python
# Don't worry about understanding this code!
# It's just to explain how KDE works step by step

import numpy as np  # for numerical operations
import matplotlib.pyplot as plt  # for plotting
from scipy import stats  # for statistical functions

# Create dataset
dataset = np.random.randn(25)
# generate 25 random values from normal distribution

# Create rugplot
sns.rugplot(dataset)
# show small lines for each data point on x-axis

# Set up the x-axis limits
x_min = dataset.min() - 2  # little space before minimum
x_max = dataset.max() + 2  # little space after maximum

# Create 100 evenly spaced points between x_min and x_max
x_axis = np.linspace(x_min, x_max, 100)

# Bandwidth calculation (controls smoothness of KDE)
bandwidth = ((4 * dataset.std()**5) / (3 * len(dataset)))**.2

# Create an empty list to store kernels
kernel_list = []

# Plot each small normal curve (kernel) for every data point
for data_point in dataset:

    # Create normal distribution centered at each data point
    kernel = stats.norm(data_point, bandwidth).pdf(x_axis)
    kernel_list.append(kernel)

    # Scale it down just for visualization
    kernel = kernel / kernel.max()
    kernel = kernel * 0.4

    # Plot the small curve
    plt.plot(x_axis, kernel, color='grey', alpha=0.5)

plt.ylim(0,1)  # set y-axis limit for better view
```

```python
# To get the KDE curve we sum all the small normal curves (kernels)

# Sum all kernel functions together
sum_of_kde = np.sum(kernel_list, axis=0)
# axis=0 means sum vertically (point by point)

# Plot the final KDE curve
fig = plt.plot(x_axis, sum_of_kde, color='indianred')
# plot the smooth density curve

# Add the rugplot again
sns.rugplot(dataset, c='indianred')
# show original data points under the curve

# Remove y-axis ticks (just for cleaner look)
plt.yticks([])

# Set title
plt.suptitle("Sum of the Basis Functions")
# add a title to the whole figure
```

```python
import seaborn as sns  # import seaborn library for visualization
%matplotlib inline
# show plots directly inside Jupyter Notebook
```

```python
tips = sns.load_dataset('tips')
# load the built-in "tips" dataset from seaborn into variable tips
```

```python
tips.head()  # display the first 5 rows of the tips dataset
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips)
# create a scatter plot with a regression line
# shows relationship between total_bill and tip
# lmplot automatically adds the best-fit line
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips, hue='sex')
# create scatter plot with regression lines
# hue='sex' separates data by male and female
# each group gets its own color and regression line
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips, hue='sex', palette='coolwarm')
# create scatter plot with regression lines
# hue='sex' separates male and female
# palette='coolwarm' changes the color style
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips, col='sex')
# create separate regression plots for each sex
# col='sex' makes one plot for male and one for female
```

```python
sns.lmplot(x="total_bill", y="tip", row="sex", col="time", data=tips)
# create multiple regression plots
# row="sex" separates plots by male and female (rows)
# col="time" separates plots by lunch and dinner (columns)
# shows relationship between total_bill and tip for each group
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips,
           col='day', hue='sex', palette='coolwarm')
# create separate regression plots for each day
# hue='sex' colors male and female differently
# palette='coolwarm' sets the color style
# shows total_bill vs tip relationship for each day and sex
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips,
           col='day', hue='sex', palette='coolwarm',
           aspect=0.6, height=8)
# create separate regression plots for each day
# hue='sex' colors male and female differently
# aspect=0.6 controls width-to-height ratio of each plot
# height=8 sets the height (size) of each subplot
# makes the plots taller and more compact
```
